# Run PRCA Survey with LLM Responder
This notebook allows you to easily run the `SurveyResponder` using the PRCA questions. Just run the cells below from top to bottom.

**Prerequisites:**
- Ensure you have Ollama installed and running on your machine.
- Ensure you have pulled the required model. By default, this uses `llama3.1:latest`.
  *(If you haven't yet, open a terminal and run `ollama pull llama3.1`)*

```python
from SurveyResponder import SurveyResponder
import pandas as pd
```

```python
# Initialize the Survey Responder
# You can adjust the num_responses, temperature, and model_name as needed.
responder = SurveyResponder(
    questions_path="prca_questions.json",
    persona_path="persona.json",
    model_name="llama3.1:latest",
    num_responses=10,  # Number of simulated respondents to generate
    temperature=1.0
)

# Run the survey and save the results to a CSV file
output_file = "prca_results.csv"
print(f"Generating responses and saving to {output_file}...")

# This might take a few minutes depending on the number of responses and your machine's speed
responder.run_write(output_file)
print("Done generating responses!")
```

```python
# Load and view the generated results
df = pd.read_csv(output_file)

# Display the first few rows of the generated synthetic survey data
display(df.head())
```
